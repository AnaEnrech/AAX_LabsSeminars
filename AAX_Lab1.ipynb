```python
import matplotlib.pyplot as plt

# Positions
nodes = {
    "N0": (0, 0),   # AP_A
    "N1": (0, 4),   # STA_A1
    "N2": (5, 0),   # AP_B
    "N3": (5, 4),   # STA_B1
}

# Matrix of power strength (dBm)
power_matrix = {
    ("N0", "N1"): -69.65,
    ("N0", "N2"): -77.11,
    ("N0", "N3"): -87.47,

    ("N1", "N0"): -69.65,
    ("N1", "N2"): -87.47,
    ("N1", "N3"): -77.11,

    ("N2", "N0"): -77.11,
    ("N2", "N1"): -87.47,
    ("N2", "N3"): -69.64,

    ("N3", "N0"): -87.47,
    ("N3", "N1"): -77.11,
    ("N3", "N2"): -69.64,
}

# Draw nodes
for name, (x, y) in nodes.items():
    if name in ["N0", "N2"]:  # APs
        plt.scatter(x, y, color='red', s=100, label="AP" if "AP" not in plt.gca().get_legend_handles_labels()[1] else "")
    else:  # STAs
        plt.scatter(x, y, color='blue', s=100, label="STA" if "STA" not in plt.gca().get_legend_handles_labels()[1] else "")

    plt.text(x + 0.1, y + 0.1, name)

# Dibujar enlaces SIN duplicados
drawn = set()

for (n1, n2), power in power_matrix.items():
    if (n2, n1) in drawn:
        continue

    x1, y1 = nodes[n1]
    x2, y2 = nodes[n2]

    # Line
    plt.plot([x1, x2], [y1, y2], color='black', linewidth=1)


    xm = (x1 + x2) / 2
    ym = (y1 + y2) / 2

    # Label
    plt.text(xm, ym + 0.2, f"{power:.2f} dBm", fontsize=8, ha='center')

    drawn.add((n1, n2))


plt.title("Network Scenario (Channel 0)")
plt.xlabel("X")
plt.ylabel("Y")
plt.legend()
plt.grid()

plt.show()
```

For this exercise, we have applied the algorithm "Win Stay, Lose Shift" which stays in the channel if the reward received is good enough (if it's above the threshold) and changes to another one (randomly) if it isn't.

To implement it, we save the last action applied and check its reward. This algorithm works well in Wi-Fi because it reacts quickly to performance drops.

The implemented code in C++ that we included in the file multi_armed_bandits.h is as follows:

```python
# (response to exercise 4)
int PickArmML4Net(int num_arms, double *average_reward_per_arm, int *available_arms) {

    static int last_action = -1;

    //we define the threshold
    double threshold = 0.5;

    int action_ix = 0;

    //initial state
    if (last_action == -1) {
        //we choose a channel randomly
        do {
            action_ix = rand() % num_arms;
        } while (!available_arms[action_ix]);

        last_action = action_ix;
        return action_ix;
    }

    //we check the reward from the step applied
    if (average_reward_per_arm[last_action] >= threshold) { //if it's above the threshold, we stay

        action_ix = last_action;

    } else { //if it's under the threshold, we change the channel
        int attempts = 0; //we save the number of attempts to find a channel available
        do {
            action_ix = rand() % num_arms;
            attempts++;

            //if there's only one channel, we stay in that one
            if (attempts > 100) {
                action_ix = last_action;
                break;
            }
        //we repeat the action if the channel isn't available or if it's the same channel that has failed
        } while (!available_arms[action_ix] || action_ix == last_action);
    }

    //we save the action choosen
    last_action = action_ix;

    return action_ix;
}
```

To get all the results of the average throughput from all simulations, we have used a parser function.

```python
#parser

import re
import os

def parse_throughput(filepath: str) -> list[float]:
    if not os.path.exists(filepath):
        raise FileNotFoundError(f"No se encontró el archivo: {filepath}")

    throughput_values = []
    pattern = re.compile(
        r'(?:average|avg)\s+throughput\s*[=:]\s*([\d]+(?:[.,][\d]+)?)',
        re.IGNORECASE
    )

    with open(filepath, 'r', errors='replace') as f:
        for line_num, line in enumerate(f, start=1):
            match = pattern.search(line)
            if match:
                value_str = match.group(1).replace(',', '.')
                throughput_values.append(float(value_str))

    return throughput_values
```
